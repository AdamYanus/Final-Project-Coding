## Does the amount of Hits have any relation to the amount of Walks? 

```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy
```

```python
hitter_data = pd.read_csv("Hitters.csv.xls")
hitter_data
```

```python
hitter_data.info()
```

```python
hits = hitter_data["Hits"]
walks = hitter_data["Walks"]
```

```python
corr = scipy.stats.pearsonr( hits, walks)
print(corr)

correlation_score = corr[0]
print("The Correlation Score is:", correlation_score)
# Correlation score of Hits and Walks
```

```python
linear_model = scipy.stats.linregress(hits, walks)

print(linear_model)
```

```python
slope = linear_model[0]
print(slope)
```

```python
y_intercept = linear_model.intercept
```

```python
linear_fit = slope*hits+y_intercept
```

```python
plt.scatter(hits, walks, s=10,label="Hits")
plt.plot(hits, linear_fit, color="red", label=("LinearFit"))
plt.xlabel("Number of Hits")
plt.ylabel("Number of Walks")
plt.legend()
```
